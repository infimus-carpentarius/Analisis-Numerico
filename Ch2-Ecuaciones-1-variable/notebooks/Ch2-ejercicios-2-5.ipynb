## Ejercicio 13 – Aceleración de una sucesión linealmente convergente

**Enunciado:**  
Considere la sucesión $p_n = \dfrac{1}{n}$, $n \ge 1$, que converge linealmente a $0$.  
Aplique el método $\Delta^2$ de Aitken para generar la sucesión acelerada $\hat{p}_n$ y determine cuántos términos se necesitan para que $|p_n| \le 5\times10^{-m}$ y $|\hat{p}_n| \le 5\times10^{-m}$ para $m = 2, 5, 8, 10$.

**Solución teórica:**

Para la sucesión original $p_n = 1/n$, la condición $|p_n| \le \varepsilon$ se cumple cuando $n \ge 1/\varepsilon$.  
Por tanto:
- $\varepsilon = 5\times10^{-2}$ → $n \ge 20$
- $\varepsilon = 5\times10^{-5}$ → $n \ge 20\,000$
- $\varepsilon = 5\times10^{-8}$ → $n \ge 20\,000\,000$
- $\varepsilon = 5\times10^{-10}$ → $n \ge 2\times10^{9}$

La sucesión acelerada $\hat{p}_n$ se obtiene aplicando la fórmula de Aitken a tripletes consecutivos:
$$ \hat{p}_n = p_n - \frac{(p_{n+1}-p_n)^2}{p_{n+2} - 2p_{n+1} + p_n}. $$

Para $p_n = 1/n$, se puede demostrar que $\hat{p}_n = \dfrac{1}{n+1}$ (¡exactamente!). Esto significa que la aceleración produce el término siguiente de la sucesión original, por lo que la convergencia no mejora realmente (sigue siendo lineal). Sin embargo, en la práctica, con errores de redondeo, puede haber una ligera mejora. Para este ejercicio, consideraremos el comportamiento teórico y también realizaremos una verificación numérica.

**Resultados numéricos:**  
Usando un pequeño programa, calculamos el menor índice $n$ tal que $|p_n| \le \varepsilon$ y el menor índice $n$ tal que $|\hat{p}_n| \le \varepsilon$. Los resultados se muestran en la tabla.

```python
import math

def aitken(p0, p1, p2):
    """Un paso del método Δ² de Aitken."""
    num = (p1 - p0) ** 2
    den = p2 - 2*p1 + p0
    if abs(den) < 1e-15:
        return p2
    return p0 - num / den

def generar_sucesion_acelerada(n_max):
    """Genera la sucesión acelerada a partir de p_n = 1/n hasta n_max+2."""
    p = [1/(i+1) for i in range(n_max+3)]  # p[0]=1, p[1]=1/2, ...
    p_hat = []
    for n in range(n_max):
        p_hat.append(aitken(p[n], p[n+1], p[n+2]))
    return p_hat

precisiones = [5e-2, 5e-5, 5e-8]
print("Ejercicio 13 – Sucesión p_n = 1/n")
print("ε\t\tOriginal (n)\tAcelerada (n)")
for eps in precisiones:
    # Original
    n_orig = math.ceil(1/eps)
    # Acelerada: buscar el primer índice tal que |p_hat[n]| <= eps
    # Para p_n = 1/n, teóricamente p_hat[n] = 1/(n+2)? No exactamente.
    # Calculamos numéricamente:
    n_acc = 1
    while True:
        p0, p1, p2 = 1/n_acc, 1/(n_acc+1), 1/(n_acc+2)
        p_hat = aitken(p0, p1, p2)
        if abs(p_hat) <= eps:
            break
        n_acc += 1
    print(f"{eps:.0e}\t\t{n_orig}\t\t{n_acc}")
```

## Ejercicio 14 – Convergencia superlineal

**Enunciado:**  
Se dice que una sucesión $\{p_n\}$ es **superlinealmente convergente** a $p$ si  

$$ \lim_{n\to\infty} \frac{|p_{n+1} - p|}{|p_n - p|} = 0. $$

a) Muestre que si $p_n \to p$ de orden $\alpha > 1$, entonces $\{p_n\}$ es superlinealmente convergente a $p$.  
b) Muestre que $p_n = \dfrac{1}{n}$ es superlinealmente convergente a $0$ pero no converge a $0$ de orden $\alpha$ para ningún $\alpha > 1$.

---

### Solución (a)

Si $\{p_n\}$ converge a $p$ con orden $\alpha > 1$, por definición existe una constante $\lambda > 0$ tal que  

$$ \lim_{n\to\infty} \frac{|p_{n+1} - p|}{|p_n - p|^\alpha} = \lambda. $$

Entonces podemos escribir  

$$ \frac{|p_{n+1} - p|}{|p_n - p|} = \frac{|p_{n+1} - p|}{|p_n - p|^\alpha} \cdot |p_n - p|^{\alpha-1}. $$

Tomando límite cuando $n\to\infty$, como $\alpha-1 > 0$ y $|p_n - p| \to 0$, se tiene  

$$ \lim_{n\to\infty} \frac{|p_{n+1} - p|}{|p_n - p|} = \lambda \cdot 0 = 0. $$

Por tanto, $\{p_n\}$ es superlinealmente convergente.

---

### Solución (b)

Para $p_n = 1/n$, el límite es $p = 0$. Calculamos el cociente de errores sucesivos:

$$ \frac{|p_{n+1} - 0|}{|p_n - 0|} = \frac{1/(n+1)}{1/n} = \frac{n}{n+1} \to 1 \quad \text{cuando } n\to\infty. $$

Como el límite es $1$ y no $0$, la sucesión **no** es superlinealmente convergente. De hecho, el límite es $1$, lo que indica convergencia lineal con constante asintótica $\lambda = 1$.

Ahora veamos si converge con orden $\alpha > 1$. Supongamos que existiera $\alpha > 1$ tal que  

$$ \lim_{n\to\infty} \frac{|p_{n+1}|}{|p_n|^\alpha} = \lambda \quad (0<\lambda<\infty). $$

Sustituyendo $p_n = 1/n$:

$$ \frac{1/(n+1)}{(1/n)^\alpha} = \frac{n^\alpha}{n+1} = n^{\alpha-1} \cdot \frac{n}{n+1}. $$

Para $\alpha > 1$, $n^{\alpha-1} \to \infty$, por lo que el cociente tiende a infinito, no a una constante finita. Para $\alpha = 1$, el cociente tiende a $1$ (convergencia lineal). Para $\alpha < 1$, el cociente tiende a $0$, pero $\alpha$ debe ser mayor que $1$ por definición de orden de convergencia (normalmente $\alpha \ge 1$). Por tanto, no existe $\alpha > 1$ que dé un límite finito y no nulo. Así, $\{1/n\}$ no tiene orden de convergencia $\alpha > 1$.

---

### Observación

El concepto de convergencia superlineal es más débil que la convergencia de orden $\alpha > 1$; una sucesión puede ser superlineal sin tener un orden $\alpha$ fijo (por ejemplo, $p_n = 1/n!$). El caso $p_n = 1/n$ no es superlineal porque el cociente tiende a $1$, no a $0$.

## Ejercicio 15 (sección 2.5) – Relación entre convergencia superlineal y diferencia de términos consecutivos

**Enunciado:**  
Suponga que $\{p_n\}$ es superlinealmente convergente a $p$. Muestre que  
\[
\lim_{n\to\infty} \frac{|p_{n+1} - p_n|}{|p_n - p|} = 1.
\]

---

### Demostración

Por definición de convergencia superlineal,  
\[
\lim_{n\to\infty} \frac{|p_{n+1} - p|}{|p_n - p|} = 0.
\]

Escribimos  
\[
p_{n+1} - p_n = (p_{n+1} - p) - (p_n - p).
\]

Aplicando la desigualdad triangular inversa y la triangular directa:  
\[
\bigl| |p_{n+1} - p| - |p_n - p| \bigr| \le |p_{n+1} - p_n| \le |p_{n+1} - p| + |p_n - p|.
\]

Dividimos por $|p_n - p|$ (para $n$ suficientemente grande, $p_n \neq p$):  
\[
\left| \frac{|p_{n+1} - p|}{|p_n - p|} - 1 \right| \le \frac{|p_{n+1} - p_n|}{|p_n - p|} \le \frac{|p_{n+1} - p|}{|p_n - p|} + 1.
\]

Tomamos límite cuando $n\to\infty$. Por hipótesis, $\frac{|p_{n+1} - p|}{|p_n - p|} \to 0$. Entonces el lado izquierdo tiende a $|0-1| = 1$ y el lado derecho tiende a $0+1 = 1$. Por el teorema del sándwich, concluimos que  
\[
\lim_{n\to\infty} \frac{|p_{n+1} - p_n|}{|p_n - p|} = 1.
\]

## Ejercicio 16 (sección 2.5) – Demostración del teorema 2.14 (Aitken)

**Enunciado:**  
Pruebe el teorema 2.14:  
Suponga que $\{p_n\}$ es una sucesión que converge linealmente a $p$ y que  
\[
\lim_{n\to\infty} \frac{p_{n+1} - p}{p_n - p} = \lambda \quad \text{con } |\lambda| < 1.
\]  
Entonces la sucesión $\{\hat{p}_n\}$ definida por el método $\Delta^2$ de Aitken,  
\[
\hat{p}_n = p_n - \frac{(p_{n+1} - p_n)^2}{p_{n+2} - 2p_{n+1} + p_n},
\]  
converge a $p$ más rápido que $\{p_n\}$ en el sentido de que  
\[
\lim_{n\to\infty} \frac{\hat{p}_n - p}{p_n - p} = 0.
\]

---

### Demostración

Definimos el error $e_n = p_n - p$. Por hipótesis, $e_n \neq 0$ para $n$ grande y  
\[
\lim_{n\to\infty} \frac{e_{n+1}}{e_n} = \lambda, \quad |\lambda| < 1.
\]

Podemos escribir $e_{n+1} = (\lambda + \delta_n) e_n$, donde $\delta_n \to 0$.  
Entonces  
\[
p_{n+1} - p_n = e_{n+1} - e_n = (\lambda + \delta_n - 1) e_n.
\]

Además,  
\[
p_{n+2} - 2p_{n+1} + p_n = e_{n+2} - 2e_{n+1} + e_n.
\]

Usando la relación recurrente,  
\[
e_{n+2} = (\lambda + \delta_{n+1}) e_{n+1} = (\lambda + \delta_{n+1})(\lambda + \delta_n) e_n.
\]

Sustituimos en la expresión del denominador:  
\[
e_{n+2} - 2e_{n+1} + e_n = \bigl[(\lambda + \delta_{n+1})(\lambda + \delta_n) - 2(\lambda + \delta_n) + 1\bigr] e_n.
\]

Desarrollamos el corchete:  
\[
(\lambda^2 + \lambda\delta_n + \lambda\delta_{n+1} + \delta_n\delta_{n+1}) - 2\lambda - 2\delta_n + 1.
\]

Agrupamos términos constantes en $\lambda$: $\lambda^2 - 2\lambda + 1 = (1 - \lambda)^2$. Los términos que contienen $\delta_n$ o $\delta_{n+1}$ tienden a $0$. Por tanto,  
\[
p_{n+2} - 2p_{n+1} + p_n = \bigl[(1 - \lambda)^2 + \varepsilon_n\bigr] e_n,
\]  
donde $\varepsilon_n \to 0$.

Ahora calculamos el numerador de la fórmula de Aitken:  
\[
(p_{n+1} - p_n)^2 = \bigl[(\lambda + \delta_n - 1) e_n\bigr]^2 = (1 - \lambda - \delta_n)^2 e_n^2.
\]

Entonces,  
\[
\hat{p}_n - p = e_n - \frac{(p_{n+1} - p_n)^2}{p_{n+2} - 2p_{n+1} + p_n}
= e_n - \frac{(1 - \lambda - \delta_n)^2 e_n^2}{\bigl[(1 - \lambda)^2 + \varepsilon_n\bigr] e_n}
= e_n \left[ 1 - \frac{(1 - \lambda - \delta_n)^2}{(1 - \lambda)^2 + \varepsilon_n} \right].
\]

Simplificamos el corchete:  
\[
1 - \frac{(1 - \lambda - \delta_n)^2}{(1 - \lambda)^2 + \varepsilon_n}
= \frac{(1 - \lambda)^2 + \varepsilon_n - (1 - \lambda - \delta_n)^2}{(1 - \lambda)^2 + \varepsilon_n}.
\]

Desarrollamos el numerador:  
\[
(1 - \lambda)^2 + \varepsilon_n - \bigl[(1 - \lambda)^2 - 2(1 - \lambda)\delta_n + \delta_n^2\bigr] = \varepsilon_n + 2(1 - \lambda)\delta_n - \delta_n^2.
\]

Por tanto,  
\[
\hat{p}_n - p = e_n \cdot \frac{\varepsilon_n + 2(1 - \lambda)\delta_n - \delta_n^2}{(1 - \lambda)^2 + \varepsilon_n}.
\]

Dividiendo por $e_n$ (no nulo):  
\[
\frac{\hat{p}_n - p}{e_n} = \frac{\varepsilon_n + 2(1 - \lambda)\delta_n - \delta_n^2}{(1 - \lambda)^2 + \varepsilon_n}.
\]

Tomando límite cuando $n\to\infty$, como $\varepsilon_n \to 0$ y $\delta_n \to 0$, el numerador tiende a $0$ y el denominador tiende a $(1 - \lambda)^2 \neq 0$. Por tanto,  
\[
\lim_{n\to\infty} \frac{\hat{p}_n - p}{p_n - p} = 0.
\]

Esto demuestra que la sucesión acelerada $\{\hat{p}_n\}$ converge a $p$ más rápidamente que la original $\{p_n\}$ (el error relativo tiende a cero). $\square$

---

**Comentario:** La demostración anterior es una versión rigurosa que utiliza la expansión asintótica. En la práctica, el método $\Delta^2$ de Aitken convierte una sucesión linealmente convergente en otra con convergencia cuadrática (bajo ciertas condiciones), como se ilustra en los ejemplos numéricos.

## Ejercicio 15 – Relación entre convergencia superlineal y diferencia de términos consecutivos

**Enunciado:**  
Suponga que {p_n} es superlinealmente convergente a p. Muestre que  

$$ \lim_{n\to\infty} \frac{|p_{n+1} - p_n|}{|p_n - p|} = 1. $$

---

### Demostración

Por definición de convergencia superlineal,  

$$ \lim_{n\to\infty} \frac{|p_{n+1} - p|}{|p_n - p|} = 0. $$

Escribimos  

$$ p_{n+1} - p_n = (p_{n+1} - p) - (p_n - p). $$

Por la desigualdad triangular inversa y la triangular directa,  

$$ \bigl| |p_{n+1} - p| - |p_n - p| \bigr| \le |p_{n+1} - p_n| \le |p_{n+1} - p| + |p_n - p|. $$

Dividimos por |p_n - p| (para n suficientemente grande, p_n ≠ p):  

$$ \left| \frac{|p_{n+1} - p|}{|p_n - p|} - 1 \right| \le \frac{|p_{n+1} - p_n|}{|p_n - p|} \le \frac{|p_{n+1} - p|}{|p_n - p|} + 1. $$

Tomamos límite cuando n → ∞. Por hipótesis,  

$$ \frac{|p_{n+1} - p|}{|p_n - p|} \to 0. $$

Entonces el lado izquierdo tiende a |0-1| = 1 y el lado derecho tiende a 0+1 = 1. Por el teorema del sándwich, concluimos que  

$$ \lim_{n\to\infty} \frac{|p_{n+1} - p_n|}{|p_n - p|} = 1. $$

## Ejercicio 16 – Demostración del teorema 2.14 (Aitken)

**Enunciado:**  
Pruebe el teorema 2.14:  
Suponga que {p_n} es una sucesión que converge linealmente a p y que  

$$ \lim_{n\to\infty} \frac{p_{n+1} - p}{p_n - p} = \lambda \quad \text{con } |\lambda| < 1. $$

Entonces la sucesión {p̂_n} definida por el método Δ² de Aitken,  

$$ \hat{p}_n = p_n - \frac{(p_{n+1} - p_n)^2}{p_{n+2} - 2p_{n+1} + p_n}, $$

converge a p más rápido que {p_n} en el sentido de que  

$$ \lim_{n\to\infty} \frac{\hat{p}_n - p}{p_n - p} = 0. $$

---

### Demostración

Definimos el error e_n = p_n - p. Por hipótesis, e_n ≠ 0 para n grande y  

$$ \lim_{n\to\infty} \frac{e_{n+1}}{e_n} = \lambda, \quad |\lambda| < 1. $$

Podemos escribir e_{n+1} = (λ + δ_n) e_n, donde δ_n → 0.  
Entonces  

$$ p_{n+1} - p_n = e_{n+1} - e_n = (λ + δ_n - 1) e_n. $$

Además,  

$$ p_{n+2} - 2p_{n+1} + p_n = e_{n+2} - 2e_{n+1} + e_n. $$

Usando la relación recurrente,  

$$ e_{n+2} = (λ + δ_{n+1}) e_{n+1} = (λ + δ_{n+1})(λ + δ_n) e_n. $$

Sustituimos en la expresión del denominador:  

$$ e_{n+2} - 2e_{n+1} + e_n = \bigl[(λ + δ_{n+1})(λ + δ_n) - 2(λ + δ_n) + 1\bigr] e_n. $$

Desarrollamos el corchete:  

$$ (λ^2 + λδ_n + λδ_{n+1} + δ_nδ_{n+1}) - 2λ - 2δ_n + 1. $$

Agrupamos términos constantes en λ: λ^2 - 2λ + 1 = (1 - λ)^2. Los términos que contienen δ_n o δ_{n+1} tienden a 0. Por tanto,  

$$ p_{n+2} - 2p_{n+1} + p_n = \bigl[(1 - λ)^2 + ε_n\bigr] e_n, $$  

donde ε_n → 0.

Ahora calculamos el numerador de la fórmula de Aitken:  

$$ (p_{n+1} - p_n)^2 = \bigl[(λ + δ_n - 1) e_n\bigr]^2 = (1 - λ - δ_n)^2 e_n^2. $$

Entonces,  

$$ \hat{p}_n - p = e_n - \frac{(p_{n+1} - p_n)^2}{p_{n+2} - 2p_{n+1} + p_n}
= e_n - \frac{(1 - λ - δ_n)^2 e_n^2}{\bigl[(1 - λ)^2 + ε_n\bigr] e_n}
= e_n \left[ 1 - \frac{(1 - λ - δ_n)^2}{(1 - λ)^2 + ε_n} \right]. $$

Simplificamos el corchete:  

$$ 1 - \frac{(1 - λ - δ_n)^2}{(1 - λ)^2 + ε_n}
= \frac{(1 - λ)^2 + ε_n - (1 - λ - δ_n)^2}{(1 - λ)^2 + ε_n}. $$

Desarrollamos el numerador:  

$$ (1 - λ)^2 + ε_n - \bigl[(1 - λ)^2 - 2(1 - λ)δ_n + δ_n^2\bigr]
= ε_n + 2(1 - λ)δ_n - δ_n^2. $$

Por tanto,  

$$ \hat{p}_n - p = e_n \cdot \frac{ε_n + 2(1 - λ)δ_n - δ_n^2}{(1 - λ)^2 + ε_n}. $$

Dividiendo por e_n (no nulo):  

$$ \frac{\hat{p}_n - p}{e_n} = \frac{ε_n + 2(1 - λ)δ_n - δ_n^2}{(1 - λ)^2 + ε_n}. $$

Tomando límite cuando n → ∞, como ε_n → 0 y δ_n → 0, el numerador tiende a 0 y el denominador tiende a (1 - λ)^2 ≠ 0. Por tanto,  

$$ \lim_{n\to\infty} \frac{\hat{p}_n - p}{p_n - p} = 0. $$

Esto demuestra que la sucesión acelerada {p̂_n} converge a p más rápidamente que la original {p_n} (el error relativo tiende a cero). □

**Comentario:** La demostración anterior es una versión rigurosa que utiliza la expansión asintótica. En la práctica, el método Δ² de Aitken convierte una sucesión linealmente convergente en otra con convergencia cuadrática (bajo ciertas condiciones), como se ilustra en los ejemplos numéricos.
