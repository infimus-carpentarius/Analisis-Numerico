## Problema 5a – Método de Newton

**Función y derivada:**
f(x) = x³ - 2x² - 5
f'(x) = 3x² - 4x

**Intervalo:** [1, 4]

Evaluamos en los extremos:
- f(1) = 1 - 2 - 5 = -6 (negativo)
- f(4) = 64 - 32 - 5 = 27 (positivo)
→ existe al menos una raíz.

**Aproximaciones iniciales probadas:**
- Extremo izquierdo p₀ = 1
- Punto medio p₀ = 2.5
- Extremo derecho p₀ = 4

La raíz real está cerca de x ≈ 2.690647. Se espera que Newton converja rápidamente (convergencia cuadrática) desde cualquier punto.

```python
import math
from src.root_finding import newton

def f_a(x):
    return x**3 - 2*x**2 - 5

def df_a(x):
    return 3*x**2 - 4*x

intervalo_a = (1, 4)
puntos_a = [1.0, 2.5, 4.0]
tol = 1e-4

print("Apartado a) x^3 - 2x^2 - 5 = 0")
for p0 in puntos_a:
    raiz, hist, razon = newton(f_a, df_a, p0, tol_abs=tol, max_iter=50, verbose=False)
    print(f"p0 = {p0:4.1f} -> raíz = {raiz:.8f}, iteraciones = {len(hist)-1}, razón: {razon}")
```

**Función y derivada:**
f(x) = x³ + 3x² - 1
f'(x) = 3x² + 6x

**Intervalo:** [-3, -2]

Evaluamos:

```python
def f_b(x):
    return x**3 + 3*x**2 - 1

def df_b(x):
    return 3*x**2 + 6*x

intervalo_b = (-3, -2)
puntos_b = [-3.0, -2.5, -2.0]

print("\nApartado b) x^3 + 3x^2 - 1 = 0")
for p0 in puntos_b:
    try:
        raiz, hist, razon = newton(f_b, df_b, p0, tol_abs=tol, max_iter=50, verbose=False)
        print(f"p0 = {p0:4.1f} -> raíz = {raiz:.8f}, iteraciones = {len(hist)-1}, razón: {razon}")
    except Exception as e:
        print(f"p0 = {p0:4.1f} -> Error: {e}")
```

## Problema 5c – Método de Newton

**Función y derivada:**
f(x) = x - cos x
f'(x) = 1 + sen x

**Intervalo:** [0, π/2] ≈ [0, 1.5708]

Evaluamos:
- f(0) = 0 - 1 = -1 (negativo)
- f(π/2) = 1.5708 - 0 = 1.5708 (positivo)
→ existe una raíz única.

- Extremo izquierdo p₀ = 0
- Punto medio p₀ = π/4 ≈ 0.785398
- Extremo derecho p₀ = π/2 ≈ 1.5708

La derivada nunca se anula (1 + sen x > 0). La raíz exacta es x ≈ 0.739085.

```python
def f_c(x):
    return x - math.cos(x)

def df_c(x):
    return 1 + math.sin(x)

intervalo_c = (0, math.pi/2)
puntos_c = [0.0, math.pi/4, math.pi/2]

print("\nApartado c) x - cos x = 0")
for p0 in puntos_c:
    raiz, hist, razon = newton(f_c, df_c, p0, tol_abs=tol, max_iter=50, verbose=False)
    print(f"p0 = {p0:6.4f} -> raíz = {raiz:.8f}, iteraciones = {len(hist)-1}, razón: {razon}")
```

## Ejercicio 11a – Métodos para \(e^x + 2^{-x} + 2\cos x - 6 = 0\)

**Función:** \(f(x) = e^x + 2^{-x} + 2\cos x - 6\)

**Derivada:** \(f'(x) = e^x - 2^{-x}\ln 2 - 2\sin x\)

**Intervalo:** \([1, 2]\)  
- \(f(1) = e + 0.5 + 2\cos1 - 6 \approx 2.7183 + 0.5 + 1.0806 - 6 = -1.7011\) (negativo)  
- \(f(2) = e^2 + 0.25 + 2\cos2 - 6 \approx 7.3891 + 0.25 - 0.8323 - 6 = 0.8068\) (positivo)  
→ existe una raíz.

**Aproximaciones para cada método:**  
- **Newton:** \(p_0 = 1.5\) (punto medio)  
- **Secante:** \(p_0 = 1\), \(p_1 = 2\) (extremos)  
- **Posición falsa:** \(a = 1\), \(b = 2\)

**Tolerancia:** \(10^{-5}\)

Se espera que los tres métodos converjan a la raíz \(p \approx 1.3456\).

```python
import math
from src.root_finding import newton, secante, posicion_falsa

# Definición de la función y su derivada
def f11a(x):
    return math.exp(x) + 2**(-x) + 2*math.cos(x) - 6

def df11a(x):
    return math.exp(x) - (2**(-x))*math.log(2) - 2*math.sin(x)

intervalo = (1, 2)
tol = 1e-5

print("Ejercicio 11a: e^x + 2^{-x} + 2 cos x - 6 = 0")

# Newton con p0 = 1.5
raiz_n, hist_n, razon_n = newton(f11a, df11a, 1.5, tol_abs=tol, verbose=False)
print(f"Newton:       raíz = {raiz_n:.8f}, iteraciones = {len(hist_n)-1}, razón: {razon_n}")

# Secante con p0=1, p1=2
raiz_s, hist_s, razon_s = secante(f11a, 1.0, 2.0, tol_abs=tol, verbose=False)
print(f"Secante:      raíz = {raiz_s:.8f}, iteraciones = {len(hist_s)-2}, razón: {razon_s}")

# Posición falsa con a=1, b=2
raiz_pf, hist_pf, razon_pf = posicion_falsa(f11a, 1.0, 2.0, tol_abs=tol, verbose=False)
print(f"Posición falsa: raíz = {raiz_pf:.8f}, iteraciones = {len(hist_pf)}, razón: {razon_pf}")
```

## Ejercicio 17 – Ceros de $f(x)=\ln(x^2+1)-e^{0.4x}\cos(\pi x)$

**Función:**  
$$ f(x) = \ln(x^2+1) - e^{0.4x}\cos(\pi x) $$

**Derivada (para Newton):**  
$$ f'(x) = \frac{2x}{x^2+1} - 0.4 e^{0.4x}\cos(\pi x) + \pi e^{0.4x}\sin(\pi x) $$

**Comportamiento:**  
- El término $\ln(x^2+1)$ crece lentamente.  
- El término $e^{0.4x}\cos(\pi x)$ oscila con amplitud creciente debido a $e^{0.4x}$.  
- Los ceros ocurren cuando $\ln(x^2+1) = e^{0.4x}\cos(\pi x)$.  

Para $x$ negativo grande, $\ln(x^2+1)$ crece como $2\ln|x|$, mientras que $e^{0.4x}$ es muy pequeño, por lo que $f(x)$ es positivo? Veremos.

**Objetivos:**  
a) Encontrar el único cero negativo (dentro de $10^{-6}$).  
b) Encontrar los cuatro ceros positivos más pequeños.  
c) Dar una aproximación inicial razonable para el $n$-ésimo cero positivo.  
d) Calcular el vigesimoquinto cero positivo.

**Estrategia:**  
- Graficar la función para localizar intervalos que contengan los ceros.  
- Usar el método de Newton (con derivada analítica) partiendo de aproximaciones iniciales adecuadas.  
- Para el apartado (c), notar que para $x$ grande, el término $\cos(\pi x)$ oscila entre $-1$ y $1$, y $e^{0.4x}$ domina. Los ceros ocurren cerca de donde $\cos(\pi x)=0$, es decir, $x \approx n + 1/2$ para $n$ entero. Así, una buena aproximación inicial para el $n$-ésimo cero positivo es $p_0 = n + 0.5$.

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from src.root_finding import newton

def f17(x):
    return math.log(x*x + 1) - math.exp(0.4*x) * math.cos(math.pi*x)

def df17(x):
    term1 = 2*x / (x*x + 1)
    term2 = -0.4 * math.exp(0.4*x) * math.cos(math.pi*x)
    term3 = math.pi * math.exp(0.4*x) * math.sin(math.pi*x)
    return term1 + term2 + term3

# ------------------------------------------------------------
# Gráfica para visualizar los primeros ceros
# ------------------------------------------------------------
x_vals = np.linspace(-2, 12, 2000)
y_vals = [f17(x) for x in x_vals]

plt.figure(figsize=(12, 6))
plt.plot(x_vals, y_vals, 'b-', linewidth=1.2)
plt.axhline(0, color='k', linestyle='--', linewidth=0.8)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('$f(x)=\\ln(x^2+1)-e^{0.4x}\\cos(\\pi x)$')
plt.grid(True, alpha=0.3)
plt.ylim(-10, 10)
plt.show()

# ------------------------------------------------------------
# a) Cero negativo único
# ------------------------------------------------------------
print("Ejercicio 17")
print("\na) Cero negativo:")
# Buscar visualmente: para x negativo, f(-0.5) ≈ ln(1.25) - e^{-0.2}cos(-π/2) = 0.2231 - 0.8187*0 = 0.2231
# f(-1) = ln(2) - e^{-0.4}cos(-π) = 0.6931 - 0.6703*(-1) = 0.6931+0.6703=1.3634
# f(-0.2) = ln(1.04)-e^{-0.08}cos(-0.2π) ≈ 0.0392 - 0.9231*0.8090 = 0.0392 - 0.7465 = -0.7073
# Cambio de signo entre -0.5 y -0.2. Usamos p0 = -0.35.
raiz_neg, hist, razon = newton(f17, df17, -0.35, tol_abs=1e-6, max_iter=50, verbose=False)
print(f"  Raíz negativa ≈ {raiz_neg:.10f}, iteraciones = {len(hist)-1}, razón: {razon}")

# ------------------------------------------------------------
# b) Cuatro ceros positivos más pequeños
# ------------------------------------------------------------
print("\nb) Cuatro ceros positivos más pequeños:")
# Estimación inicial: cerca de donde cos(πx)=0, es decir, x ≈ n + 0.5 para n=0,1,2,3?
# Pero el primer cero positivo está antes de 0.5? Observamos en gráfica: alrededor de 0.3 y luego cerca de 1.5, 2.5, 3.5, etc.
p0_list = [0.3, 1.5, 2.5, 3.5]
for i, p0 in enumerate(p0_list, 1):
    raiz, hist, razon = newton(f17, df17, p0, tol_abs=1e-6, max_iter=50, verbose=False)
    print(f"  Cero {i}: x ≈ {raiz:.10f}, iteraciones = {len(hist)-1}, razón: {razon}")

# ------------------------------------------------------------
# c) Aproximación inicial para el n-ésimo cero positivo
# ------------------------------------------------------------
print("\nc) Para el n-ésimo cero positivo (n ≥ 1):")
print("  Una buena aproximación inicial es p0 = n + 0.5, porque cos(πx) se anula allí y la función cambia de signo.")
print("  Para n pequeño hay que ajustar un poco, pero para n grande es excelente.")

# ------------------------------------------------------------
# d) Vigesimoquinto cero positivo (n=25)
# ------------------------------------------------------------
print("\nd) Vigesimoquinto cero positivo (n=25):")
p0_25 = 25 + 0.5  # 25.5
raiz_25, hist_25, razon_25 = newton(f17, df17, p0_25, tol_abs=1e-6, max_iter=50, verbose=False)
print(f"  Aprox. inicial = {p0_25}")
print(f"  Raíz ≈ {raiz_25:.10f}, iteraciones = {len(hist_25)-1}, razón: {razon_25}")
```

## Ejercicio 19 – Punto de la parábola $y=x^2$ más cercano a $(0,1)$

### Análisis matemático (solución exacta)

**Planteamiento:**  
Buscamos el punto $(x, x^2)$ sobre la parábola que minimiza la distancia euclídea al punto fijo $P_0 = (0,1)$. La distancia al cuadrado es:

$$
d(x)^2 = (x-0)^2 + (x^2-1)^2 = x^2 + (x^2-1)^2.
$$

Derivamos e igualamos a cero para encontrar el mínimo:

$$
f(x) = \frac{d}{dx}\bigl[d(x)^2\bigr] = 2x + 2(x^2-1)\cdot 2x = 2x + 4x(x^2-1) = 2x + 4x^3 - 4x = 4x^3 - 2x.
$$

Simplificando:

$$
f(x) = 2x(2x^2 - 1).
$$

Los puntos críticos se obtienen de $f(x)=0$:

- $x=0$ (da $d(0)^2 = 0^2+(0-1)^2=1$, distancia 1).
- $2x^2-1=0 \Rightarrow x = \pm\frac{1}{\sqrt{2}} \approx \pm 0.70710678$.

Evaluamos la distancia en estos puntos:

- $x = 1/\sqrt{2}$: $d^2 = (1/2) + (1/2-1)^2 = 0.5 + (-0.5)^2 = 0.5+0.25=0.75$, distancia $\sqrt{0.75}\approx0.866025$.
- $x = -1/\sqrt{2}$: mismo valor por simetría.
- $x=0$: distancia 1.

Por tanto, el mínimo se alcanza en $x = \pm 1/\sqrt{2}$. La distancia mínima es $\sqrt{0.75} = \sqrt{3}/2 \approx 0.86602540378$.

Los puntos sobre la parábola son $\left(\frac{1}{\sqrt{2}}, \frac{1}{2}\right)$ y $\left(-\frac{1}{\sqrt{2}}, \frac{1}{2}\right)$. Ambos están a la misma distancia.

### Análisis numérico (método de Newton)

Para aplicar Newton, debemos resolver $f(x)=4x^3-2x=0$. La derivada es $f'(x)=12x^2-2$. La iteración de Newton es:

$$
x_{n+1} = x_n - \frac{4x_n^3-2x_n}{12x_n^2-2}.
$$

Podemos simplificar: $x_{n+1} = x_n - \frac{2x_n(2x_n^2-1)}{2(6x_n^2-1)} = x_n - \frac{x_n(2x_n^2-1)}{6x_n^2-1}$.

Elegimos una aproximación inicial, por ejemplo $x_0 = 0.5$ (cerca de la raíz positiva). Iteramos hasta alcanzar $10^{-4}$.

## Ejercicio 21 – Determinación de dos números que cumplen una condición

**Enunciado:**  
La suma de dos números es 20. Si cada número se suma a su raíz cuadrada, el producto de las dos sumas es 155.55. Determine los dos números dentro de \(10^{-4}\).

### Análisis matemático

Sean los números \(x\) e \(y\). Tenemos:

$$
\begin{cases}
x + y = 20,\\
(x + \sqrt{x})\,(y + \sqrt{y}) = 155.55.
\end{cases}
$$

Sustituimos \(y = 20 - x\). Definimos la función:

$$
f(x) = (x + \sqrt{x})\bigl((20-x) + \sqrt{20-x}\bigr) - 155.55 = 0.
$$

El dominio es \(0 , 20\). Por simetría, las soluciones son intercambiables.

### Análisis numérico

Aplicamos el método de Newton con derivada analítica:

$$
f'(x) = \left(1 + \frac{1}{2\sqrt{x}}\right)\bigl(20 - x + \sqrt{20-x}\bigr) + \bigl(x + \sqrt{x}\bigr)\left(-1 - \frac{1}{2\sqrt{20-x}}\right).
$$

Elegimos una aproximación inicial (por ejemplo \(x_0 = 6.5\)) y alcanzamos la tolerancia \(10^{-4}\).

A continuación se muestra el código que, además de resolverlo, prueba todas las iniciales desde 0 hasta 10 con paso 0.5 para observar la convergencia.

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from src.root_finding import newton

# Definición de la función y su derivada
def f21(x):
    if x <= 0 or x >= 20:
        return float('inf')
    sqrtx = math.sqrt(x)
    sqrty = math.sqrt(20 - x)
    return (x + sqrtx) * (20 - x + sqrty) - 155.55

def df21(x):
    if x <= 0 or x >= 20:
        return float('inf')
    sqrtx = math.sqrt(x)
    sqrty = math.sqrt(20 - x)
    term1 = (1 + 0.5/sqrtx) * (20 - x + sqrty)
    term2 = (x + sqrtx) * (-1 - 0.5/sqrty)
    return term1 + term2

tol = 1e-4
print("Ejercicio 21 – Barrido de aproximaciones iniciales")
print("x0        -> raíz encontrada (iteraciones)\n")

# Lista de valores iniciales desde 0 a 10 con paso 0.5
for x0 in np.arange(0.5, 10.5, 0.5):
    try:
        raiz, hist, razon = newton(f21, df21, x0, tol_abs=tol, max_iter=50, verbose=False)
        # Mostrar con 6 decimales
        print(f"{x0:4.1f} -> {raiz:.6f} (iter={len(hist)-1})")
    except Exception as e:
        print(f"{x0:4.1f} -> Error: {e}")

# Solución con una buena inicial (por ejemplo 6.5)
print("\n--- Solución con x0 = 6.5 ---")
raiz, hist, razon = newton(f21, df21, 6.5, tol_abs=tol, max_iter=50, verbose=True)
y = 20 - raiz
print(f"\nx = {raiz:.8f}, y = {y:.8f}")
print(f"Producto: {(raiz + math.sqrt(raiz))*(y + math.sqrt(y)):.8f} (objetivo 155.55)")

# Gráfica de f(x) en el intervalo [0,20]
x_vals = np.linspace(0.01, 19.99, 500)
y_vals = [f21(x) for x in x_vals]

plt.figure(figsize=(10, 6))
# Ejes cartesianos en rojo
plt.axhline(0, color='red', linewidth=1.2, zorder=0)
plt.axvline(0, color='red', linewidth=1.2, zorder=0)
plt.plot(x_vals, y_vals, 'b-', linewidth=1.5, label='$f(x)$')
# Punto de la raíz
plt.plot(raiz, 0, 'ro', markersize=8, label=f'Raíz x≈{raiz:.4f}')
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.title('$f(x) = (x+\\sqrt{x})(20-x+\\sqrt{20-x})-155.55$')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(-1, 21)
plt.ylim(-20, 20)
plt.show()
```

## Ejercicio 22 – Determinación de la tasa de natalidad en un modelo de población

**Problema:**  
Una población inicial $N(0)=1\,000\,000$ individuos recibe una inmigración constante de $435\,000$ personas por año. Al final del primer año la población es $N(1)=1\,564\,000$.  
La ecuación que relaciona la tasa de natalidad $\lambda$ (constante) es:

$$
1\,564\,000 = 1\,000\,000\,e^{\lambda} + \frac{435\,000}{\lambda}\bigl(e^{\lambda} - 1\bigr).
$$

Se pide:

a) Encontrar $\lambda$ con precisión $10^{-4}$.  
b) Usar ese valor para predecir la población al final del segundo año (suponiendo la misma tasa de inmigración).

### Análisis matemático

Reescribimos la ecuación como:

$$
f(\lambda) = 1\,000\,000\,e^{\lambda} + \frac{435\,000}{\lambda}\bigl(e^{\lambda} - 1\bigr) - 1\,564\,000 = 0.
$$

Derivada (para aplicar el método de Newton):

$$
f'(\lambda) = 1\,000\,000\,e^{\lambda} + 435\,000\left[\frac{e^{\lambda}\lambda - (e^{\lambda}-1)}{\lambda^2}\right] = 1\,000\,000\,e^{\lambda} + \frac{435\,000}{\lambda^2}\bigl(e^{\lambda}(\lambda-1)+1\bigr).
$$

El dominio de $\lambda$ es $\lambda > 0$. Aproximación inicial: podemos estimar $\lambda$ a partir de un crecimiento simple. Si no hubiera inmigración, $\frac{1.564}{1} = e^{\lambda}$ daría $\lambda \approx \ln(1.564) \approx 0.447$. Con inmigración, la tasa será menor. Tomamos $\lambda_0 = 0.1$.

### Análisis numérico

Aplicamos el método de Newton con tolerancia $10^{-4}$. Una vez obtenido $\lambda$, la población al segundo año se calcula como:

$$
N(2) = 1\,000\,000\,e^{2\lambda} + \frac{435\,000}{\lambda}\bigl(e^{2\lambda} - 1\bigr).
$$

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from src.root_finding import newton

# Definición de la función f(λ) y su derivada
def f_lambda(lam):
    return 1_000_000 * math.exp(lam) + (435_000 / lam) * (math.exp(lam) - 1) - 1_564_000

def df_lambda(lam):
    exp_lam = math.exp(lam)
    term1 = 1_000_000 * exp_lam
    term2 = (435_000 / (lam**2)) * (exp_lam * (lam - 1) + 1)
    return term1 + term2

# Parámetros
tol = 1e-4
lam0 = 0.1

print("Ejercicio 22 – Cálculo de λ")
lam, hist, razon = newton(f_lambda, df_lambda, lam0, tol_abs=tol, max_iter=50, verbose=True)

print(f"\n--- Resultados ---")
print(f"λ = {lam:.8f}")
print(f"f(λ) = {f_lambda(lam):.2e}")
print(f"Razón: {razon}")

# Predicción de la población al final del segundo año
exp_2lam = math.exp(2*lam)
N2 = 1_000_000 * exp_2lam + (435_000 / lam) * (exp_2lam - 1)
print(f"\nPoblación estimada al final del segundo año: {N2:.0f} individuos")

# Gráfica de f(λ) en un intervalo que contenga la raíz
lam_vals = np.linspace(0.01, 0.5, 500)
f_vals = [f_lambda(lam) for lam in lam_vals]

plt.figure(figsize=(10, 6))
# Ejes cartesianos en rojo
plt.axhline(0, color='red', linewidth=1.2, zorder=0)
plt.axvline(0, color='red', linewidth=1.2, zorder=0)
plt.plot(lam_vals, f_vals, 'b-', linewidth=1.5, label='$f(\\lambda)$')
# Punto de la raíz
plt.plot(lam, 0, 'ro', markersize=8, label=f'Raíz λ≈{lam:.5f}')
plt.xlabel('λ', fontsize=12)
plt.ylabel('f(λ)', fontsize=12)
plt.title('Ecuación de población: $f(\\lambda)=0$')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(0, 0.5)
plt.ylim(-100_000, 100_000)
plt.show()
```

## Ejercicio 23 – Tasa de interés máxima en una hipoteca

**Problema:**  
La ecuación de anualidad ordinaria es:

$$
A = \frac{P}{i}\left[1 - (1+i)^{-n}\right],
$$

donde  
$A$ = monto de la hipoteca (\$135,000),  
$P$ = pago mensual (\$1,000),  
$n$ = número de periodos (30 años × 12 meses = 360 meses),  
$i$ = tasa de interés mensual (incógnita).

Se pide encontrar la tasa máxima que el prestatario puede pagar.

### Análisis matemático

Definimos la función:

$$
f(i) = \frac{P}{i}\left[1 - (1+i)^{-n}\right] - A = 0.
$$

Sustituyendo valores:

$$
f(i) = \frac{1000}{i}\left[1 - (1+i)^{-360}\right] - 135\,000.
$$

Derivada (para el método de Newton):

$$
f'(i) = 1000\,\frac{n(1+i)^{-n-1}\,i - \bigl[1 - (1+i)^{-n}\bigr]}{i^2}.
$$

### Análisis numérico

Aplicamos el método de Newton con tolerancia $10^{-8}$ (para alta precisión).  
Aproximación inicial: una tasa anual típica del 5% equivale a $i \approx 0.05/12 \approx 0.004167$. Probamos $i_0 = 0.005$.

Una vez encontrada $i$, la tasa anual efectiva es $(1+i)^{12}-1$.

La gráfica muestra la función $f(i)$ y su raíz.

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from src.root_finding import newton

# Parámetros
A = 135_000
P = 1000
n = 360   # meses

def f_mortgage(i):
    """Función f(i) = P/i * (1 - (1+i)^{-n}) - A"""
    if i <= 0:
        return float('inf')
    return (P / i) * (1 - (1 + i) ** (-n)) - A

def df_mortgage(i):
    """Derivada analítica de f(i)"""
    if i <= 0:
        return float('inf')
    factor = (1 + i) ** (-n - 1)
    numer = n * factor * i - (1 - (1 + i) ** (-n))
    return P * numer / (i * i)

# Tolerancia e inicial
tol = 1e-8
i0 = 0.005   # 0.5% mensual (aprox. 6% anual)

print("Ejercicio 23 – Cálculo de la tasa de interés máxima")
i_root, hist, razon = newton(f_mortgage, df_mortgage, i0, tol_abs=tol, max_iter=50, verbose=True)

print(f"\n--- Resultados ---")
print(f"Tasa mensual i = {i_root:.8f} ({i_root*100:.4f}% mensual)")
tasa_anual = (1 + i_root)**12 - 1
print(f"Tasa anual efectiva = {tasa_anual:.6f} ({tasa_anual*100:.4f}% anual)")

# Verificar el monto con la tasa encontrada
monto_calculado = (P / i_root) * (1 - (1 + i_root) ** (-n))
print(f"Monto de la hipoteca con esta tasa: ${monto_calculado:.2f} (objetivo ${A})")
print(f"Razón de parada: {razon}")

# Gráfica de f(i) alrededor de la raíz
i_vals = np.linspace(0.001, 0.01, 500)
f_vals = [f_mortgage(i) for i in i_vals]

plt.figure(figsize=(10, 6))
# Ejes cartesianos en rojo
plt.axhline(0, color='red', linewidth=1.2, zorder=0)
plt.axvline(0, color='red', linewidth=1.2, zorder=0)
plt.plot(i_vals, f_vals, 'b-', linewidth=1.5, label='$f(i)$')
# Raíz
plt.plot(i_root, 0, 'ro', markersize=8, label=f'Raíz i≈{i_root:.6f}')
plt.xlabel('Tasa de interés mensual $i$', fontsize=12)
plt.ylabel('$f(i)$', fontsize=12)
plt.title('Función $f(i) = \\frac{1000}{i}[1-(1+i)^{-360}] - 135000$')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(0.001, 0.01)
plt.ylim(-20000, 20000)
plt.show()
```

## Ejercicio 27 – Cálculo de la probabilidad de punto en ráquetbol

**Problema:**  
En un juego de ráquetbol, la probabilidad de que el jugador A gane un juego por 21‑0 (sin que B anote ningún punto) está dada por:

$$
P = \frac{1+p}{2}\left(\frac{p}{1-p+p^2}\right)^{21},
$$

donde $p$ es la probabilidad de que A gane un punto cuando él saca.  
Se pide determinar $p$ para un valor dado de $P$ (por ejemplo, $P=0.5$).

### Contexto estadístico

- El modelo asume que el jugador que saca tiene ventaja.
- La fórmula se deriva de la probabilidad de ganar un punto cuando se tiene el saque y de la alternancia del saque después de cada punto.
- El factor $\frac{1+p}{2}$ representa la probabilidad de que A gane el primer punto (ya que saca A al inicio del juego).
- El término $\left(\frac{p}{1-p+p^2}\right)^{21}$ es la probabilidad de que A gane los siguientes 20 puntos (para llegar a 21‑0), considerando que después de cada punto el saque cambia.

Para un $P$ dado, debemos resolver la ecuación no lineal en $p$ (con $0<p<1$). Usaremos el método de Newton.

### Función objetivo

$$
f(p) = \frac{1+p}{2}\left(\frac{p}{1-p+p^2}\right)^{21} - P = 0.
$$

**Derivada** (necesaria para Newton):  
Tomamos logaritmo para derivar más fácilmente. Sea

$$
g(p) = \ln\left(\frac{1+p}{2}\right) + 21\left[\ln p - \ln(1-p+p^2)\right] - \ln P.
$$

Entonces $f(p)=0$ es equivalente a $g(p)=0$. Derivamos:

$$
g'(p) = \frac{1}{1+p} + 21\left(\frac{1}{p} - \frac{1-2p}{1-p+p^2}\right).
$$

La iteración de Newton para $p$ es:

$$
p_{n+1} = p_n - \frac{g(p_n)}{g'(p_n)}.
$$

Aproximación inicial: si $P$ es cercano a $0.5$, entonces $p$ debería ser cercano a $0.5$ (por simetría). Tomamos $p_0=0.5$.

### Cálculo numérico

A continuación se resuelve para $P=0.5$ con tolerancia $10^{-6}$. Se muestra la convergencia y se grafica la función.

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from src.root_finding import newton

def log_g(p, P):
    """Función logarítmica g(p)=0 para evitar problemas numéricos."""
    if p <= 0 or p >= 1:
        return float('inf')
    return math.log((1+p)/2) + 21*(math.log(p) - math.log(1 - p + p*p)) - math.log(P)

def dlog_g(p):
    """Derivada de g(p)."""
    if p <= 0 or p >= 1:
        return float('inf')
    return 1/(1+p) + 21*(1/p - (1-2*p)/(1 - p + p*p))

# Parámetros
P_target = 0.5
p0 = 0.5
tol = 1e-6

print("Ejercicio 27 – Ráquetbol: probabilidad de ganar 21‑0")
# Usamos la función logarítmica con Newton
raiz, hist, razon = newton(lambda p: log_g(p, P_target), dlog_g, p0, tol_abs=tol, max_iter=50, verbose=True)

print(f"\n--- Resultado para P = {P_target} ---")
print(f"p = {raiz:.8f}")
print(f"Verificación: P calculada = {(1+raiz)/2 * (raiz/(1-raiz+raiz*raiz))**21:.8f}")
print(f"Razón de parada: {razon}")

# Gráfica de f(p) (no logarítmica) para visualizar la raíz
def f_p(p, P):
    return (1+p)/2 * (p/(1-p+p*p))**21 - P

p_vals = np.linspace(0.01, 0.99, 500)
f_vals = [f_p(p, P_target) for p in p_vals]

plt.figure(figsize=(10, 6))
# Ejes cartesianos en rojo
plt.axhline(0, color='red', linewidth=1.2, zorder=0)
plt.axvline(0, color='red', linewidth=1.2, zorder=0)
plt.plot(p_vals, f_vals, 'b-', linewidth=1.5, label='$f(p)$')
plt.plot(raiz, 0, 'ro', markersize=8, label=f'Raíz p≈{raiz:.5f}')
plt.xlabel('$p$ (probabilidad de ganar un punto al sacar)', fontsize=12)
plt.ylabel('$f(p)$', fontsize=12)
plt.title('Ecuación $f(p)=\\frac{1+p}{2}\\left(\\frac{p}{1-p+p^2}\\right)^{21} - 0.5 = 0$')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(0, 1)
plt.ylim(-0.6, 0.6)
plt.show()
```

## Ejercicio 30 – Análisis de la fórmula alternativa del método de la secante

**Enunciado:**  
La ecuación de iteración para el método de la secante se puede escribir en la forma más simple  

$$
p_n = \frac{f(p_{n-1})p_{n-2} - f(p_{n-2})p_{n-1}}{f(p_{n-1}) - f(p_{n-2})}.
$$

Explique por qué, en general, es probable que esta ecuación de iteración sea menos precisa que la proporcionada en el algoritmo 2.4 (la fórmula estándar).

### Explicación

La fórmula estándar del método de la secante (Algoritmo 2.4) es:

$$
p_n = p_{n-1} - f(p_{n-1})\,\frac{p_{n-1} - p_{n-2}}{f(p_{n-1}) - f(p_{n-2})}.
$$

Ambas expresiones son matemáticamente equivalentes (se obtienen una de la otra mediante álgebra elemental). Sin embargo, desde el punto de vista **numérico** (estabilidad y error de redondeo) la forma simplificada presenta inconvenientes:

1. **Cancelación catastrófica:**  
   En la fórmula simplificada, el numerador es una resta de dos productos: $f(p_{n-1})p_{n-2} - f(p_{n-2})p_{n-1}$. Si $p_{n-1}$ y $p_{n-2}$ están muy próximos (como ocurre al acercarse a la raíz), ambos términos pueden ser muy parecidos, provocando una pérdida significativa de dígitos significativos. Lo mismo ocurre con el denominador $f(p_{n-1}) - f(p_{n-2})$, que también se vuelve muy pequeño. La combinación de dos cancelaciones puede amplificar enormemente el error de redondeo.

2. **Propagación de errores:**  
   La forma estándar utiliza una expresión del tipo $p_{n-1} - \text{(algo)}$, donde el “algo” es una corrección pequeña. Esto es numéricamente más estable porque se resta una cantidad pequeña de una cantidad mayor, evitando la cancelación directa de dos números grandes y casi iguales.

3. **Sensibilidad a errores en los datos:**  
   Si $f(p_{n-1})$ y $f(p_{n-2})$ son muy pequeños (cerca de la raíz), el denominador en la forma simplificada puede ser del orden del error de redondeo, produciendo divisiones muy inestables. La forma estándar maneja mejor esta situación porque el factor $\frac{p_{n-1} - p_{n-2}}{f(p_{n-1}) - f(p_{n-2})}$ es una aproximación de $1/f'(\xi)$ y suele ser estable cuando la derivada no es cero.

**Conclusión:** Aunque algebraicamente equivalentes, la forma simplificada es **numéricamente menos precisa** debido a la cancelación de términos semejantes. Por ello, en la práctica se recomienda usar la fórmula del algoritmo 2.4.

## Preguntas de análisis – Sección 2.3

### 1. ¿El método de Newton converge para cualquier aproximación inicial? En caso afirmativo, ¿cuál es la razón y el orden de convergencia? ¿Qué sucede si la raíz es múltiple?

**Respuesta:**  
No, el método de Newton **no converge para cualquier aproximación inicial**. La convergencia está garantizada **localmente** cuando la aproximación inicial está suficientemente cerca de la raíz y $f'(p)\neq 0$ (teorema 2.6). Si la inicial está lejos, puede diverger, oscilar o converger a otra raíz.

- **Razón de convergencia:** Cuando converge, lo hace **cuadráticamente** ($\alpha=2$) si $f'(p)\neq 0$. Es decir, el error se reduce aproximadamente como $e_{n+1} \approx C e_n^2$.

- **Raíz múltiple:** Si $f(p)=f'(p)=0$ (multiplicidad $m>1$), la convergencia se vuelve **lineal** ($\alpha=1$) con factor $1-1/m$. Para restaurar la convergencia cuadrática se puede usar la versión modificada:
  $$ x_{n+1}=x_n - m\frac{f(x_n)}{f'(x_n)}. $$
  O bien, aplicar Newton a la función $\mu(x)=f(x)/f'(x)$, que tiene raíz simple.

### 2. Si la aproximación inicial no está suficientemente cerca de la raíz, el método de Newton puede converger a una raíz errónea. Proporcione ejemplos y justificación.

**Respuesta:**  
- **Ejemplo clásico:** $f(x)=x^3-2x+2$. Con $x_0=0$, Newton converge a la raíz $x\approx -1.769$ en lugar de la esperada $x=1$ (si se ignora la existencia de otra raíz).  
- **Oscilación:** $f(x)=x^2+1$ (sin raíces reales), con $x_0=1$ genera oscilación sin convergencia.  
- **Divergencia:** $f(x)=\arctan(x)$, con $x_0=1.5$ produce iteraciones que oscilan y no convergen (zona donde $|f'(x)|$ es pequeña y la corrección es enorme).  

**Justificación:** La iteración de Newton es un método de punto fijo con $g(x)=x-f(x)/f'(x)$. La convergencia requiere $|g'(x)|<1$ en un entorno. Si $x_0$ cae en una región donde $g$ es expansiva, no converge.

### 3. La función $f(x)=0.5x^3-6x^2+21.5x-22$ tiene un cero en $x=4$. Compare los resultados de los métodos de Newton y la secante usando $p_0=5$, $p_1=4.5$ para la secante, y $p_0=5$ para Newton.

**Respuesta:**  
Para $f(x)=0.5x^3-6x^2+21.5x-22$, se tiene $f(4)=0$, $f'(x)=1.5x^2-12x+21.5$, $f'(4)=1.5\cdot16-48+21.5=24-48+21.5=-2.5$.

- **Newton** con $x_0=5$:
  $x_1=5 - f(5)/f'(5)$. $f(5)=0.5\cdot125-6\cdot25+107.5-22=62.5-150+107.5-22=-2$, $f'(5)=1.5\cdot25-60+21.5=37.5-60+21.5=-1$, luego $x_1=5-(-2)/(-1)=5-2=3$. Siguiente iteración converge a 4 rápidamente (convergencia cuadrática).

- **Secante** con $p_0=5$, $p_1=4.5$:
  $p_2 = 4.5 - f(4.5)*(4.5-5)/(f(4.5)-f(5))$. Calculando $f(4.5)=0.5*91.125-6*20.25+96.75-22=45.5625-121.5+96.75-22=-1.1875$, $f(5)=-2$, entonces $p_2=4.5 - (-1.1875)*(-0.5)/(-1.1875+2)=4.5 - (0.59375)/(0.8125)=4.5-0.73077=3.76923$. Luego $p_3$ se acercará a 4. La secante tiene orden $\approx 1.618$, más lento que Newton. En este ejemplo Newton alcanza alta precisión en 2 iteraciones, secante requiere 3-4.

### 4. La función $f(x)=x^{1/3}$ tiene una raíz en $x=0$. Compare los resultados de los métodos de Newton y la secante con $p_0=1$, $p_1=0.5$ (secante) y $p_0=1$ (Newton).

**Respuesta:**  
Para $f(x)=x^{1/3}$, $f'(x)=\frac{1}{3}x^{-2/3}$. En $x=0$ la derivada es infinita, por lo que Newton no es aplicable (fórmula da iteraciones inestables). Si se usa, se obtiene:
$$x_{n+1}=x_n - \frac{x_n^{1/3}}{(1/3)x_n^{-2/3}} = x_n - 3x_n = -2x_n.$$
Con $x_0=1$ se obtiene $x_1=-2$, $x_2=4$, $x_3=-8$, … **diverge** (oscila y crece en magnitud).

El método de la secante:
$$x_{n+1}=x_n - f(x_n)\frac{x_n-x_{n-1}}{f(x_n)-f(x_{n-1})}.$$
Con $x_0=1$, $x_1=0.5$, los valores son positivos y la iteración puede converger lentamente a 0 (la raíz). Pero debido a la pendiente infinita en la raíz, la convergencia es muy lenta (sublineal). En la práctica, ambos métodos son inadecuados para esta función; es mejor usar bisección o un método específico para raíces con derivada infinita.

**Conclusión:**  
El método de Newton requiere $f'(p)\neq 0$ y continuidad de $f'$ para convergencia cuadrática. Si la derivada es cero o infinita, la convergencia se degrada o falla. La secante puede ser más robusta en esos casos, pero también sufre.
